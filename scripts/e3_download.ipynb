```python
import preprocessamento as pp
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Carregar universo LÍQUIDO
caminho_universo = 'e3_universo_b3_liquido.txt'
if Path(caminho_universo).exists():
    with open(caminho_universo, 'r') as f:
        tickers = [linha.strip() for linha in f if linha.strip()]
    print(f"✓ Universo líquido: {len(tickers)} papéis")
else:
    raise FileNotFoundError(f"Arquivo '{caminho_universo}' não encontrado. Certifique-se de que o arquivo existe e está no caminho correto.")    
```

```python
#excluir os tickers UGPA3.SA e PRIO3.SA da lista
tickers = [ticker for ticker in tickers if ticker not in ['UGPA3.SA', 'PRIO3.SA']]
```

```python
from datetime import date


data_base = date.today()
for ticker in tickers: 

    df = pp.montar_dataset(ticker, dias_total_historico=1500, data_base=data_base)
    df.to_csv(f'temp/{ticker} - {data_base}.csv', index=False, sep=';')
```
