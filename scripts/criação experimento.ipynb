```python
      
import mlflow
import os
from pathlib import Path

EXPERIMENT_NAME = "Modelo_B3_v2"

# 1. Defina o caminho base ABSOLUTO usando o formato padrão do seu sistema operacional.
#    Usar r'...' no Windows é uma boa prática para literais de string de caminho.
CAMINHO_BASE_MLFLOW = r'C:\palpiteiro\mlruns_v2' # Usei v2 para garantir que não há conflito com a pasta antiga


# Crie o diretório se ele não existir
os.makedirs(CAMINHO_BASE_MLFLOW, exist_ok=True)

uri_mlflow = Path(CAMINHO_BASE_MLFLOW).as_uri()


# 2. Passe o caminho do sistema de arquivos DIRETAMENTE para set_tracking_uri.
#    O MLflow irá convertê-lo internamente para o formato URI correto ('file:///...').
#    Esta é a forma recomendada e mais segura para caminhos locais.
mlflow.set_tracking_uri(uri_mlflow)


# Agora, podemos verificar qual URI ele construiu
print(f"MLflow Tracking URI definido para: {mlflow.get_tracking_uri()}")


# 3. A lógica para criar o experimento com artifact_location permanece a mesma.
#    O artifact_location também deve ser o caminho do sistema de arquivos,
#    pois mlflow.create_experiment também lida com a conversão.
try:
    experiment = mlflow.get_experiment_by_name(EXPERIMENT_NAME)
    if experiment is None:
        raise AttributeError
    
    experiment_id = experiment.experiment_id
    print(f"Experimento '{EXPERIMENT_NAME}' já existe (ID: {experiment_id}).")
    
    # Opcional: Verificar se o artifact_location do experimento existente está correto.
    # Se não estiver, pode ser necessário recriar o experimento.
    if not experiment.artifact_location.endswith(os.path.basename(CAMINHO_BASE_MLFLOW)):
         print(f"AVISO: O artifact_location do experimento existente ('{experiment.artifact_location}') "
               f"pode não apontar para o diretório desejado ('{CAMINHO_BASE_MLFLOW}'). "
               "Considere renomear/recriar o experimento se os artefatos não forem salvos corretamente.")

except AttributeError:
    print(f"Experimento '{EXPERIMENT_NAME}' não encontrado. Criando agora...")
    # Passamos o caminho do sistema de arquivos diretamente aqui também.
    experiment_id = mlflow.create_experiment(EXPERIMENT_NAME, artifact_location=uri_mlflow)
    print(f"Experimento '{EXPERIMENT_NAME}' criado com ID: {experiment_id} e Artifact Location: {uri_mlflow}")

# 4. Defina o experimento que será usado para as próximas runs
mlflow.set_experiment(experiment_name=EXPERIMENT_NAME)


# 5. Execute a run. Agora tudo deve ir para o lugar certo.
with mlflow.start_run(run_name="teste_com_uri_simplificado") as run:
    print(f"\n--- Iniciando Run ---")
    print(f"Run ID: {run.info.run_id}")
    print(f"Artifact URI para esta run: {run.info.artifact_uri}")

    features_list = [10, 20, 30]
    arquivo_features_path = "features_log.txt"
    with open(arquivo_features_path, "w") as f:
        f.write("\n".join(map(str, features_list)))
            
    mlflow.log_artifact(arquivo_features_path, artifact_path="model_config")
    print(f"Artefato '{arquivo_features_path}' logado em 'model_config'.")
    
    if os.path.exists(arquivo_features_path):
        os.remove(arquivo_features_path)

print(f"\nVerifique a pasta '{CAMINHO_BASE_MLFLOW}' e a UI do MLflow. O artefato deve estar lá.")
# Para rodar a UI:
# mlflow ui --backend-store-uri file:///C:/palpiteiro/mlruns_v2  (ou o caminho que você definiu)
# Ou, no terminal, navegue até C:\palpiteiro e rode: mlflow ui --backend-store-uri mlruns_v2

    
```

```python
import mlflow
import os

EXPERIMENT_NAME = "Modelo_B3_v3"
# 1. Defina o caminho base ABSOLUTO.
CAMINHO_BASE_MLFLOW = r'C:\palpiteiro\mlruns_v3'

# 2. Processe o caminho ANTES de passá-lo para o f-string.
#    Obtém o caminho absoluto e substitui as barras invertidas por barras normais.
caminho_processado = os.path.abspath(CAMINHO_BASE_MLFLOW).replace('\\', '/')

# 3. Agora, construa o URI com a string já processada.
tracking_uri = f"file:///{caminho_processado}"

# 4. Defina o Tracking URI
mlflow.set_tracking_uri(tracking_uri)
print(f"MLflow Tracking URI definido para: {mlflow.get_tracking_uri()}")


# 5. Verifique se o experimento existe. Se não, crie-o com o artifact_location.
try:
    experiment = mlflow.get_experiment_by_name(EXPERIMENT_NAME)
    if experiment is None:
        raise AttributeError # Força a entrada no bloco except se o experimento não existir
    
    experiment_id = experiment.experiment_id
    print(f"Experimento '{EXPERIMENT_NAME}' já existe (ID: {experiment_id}).")
    
    # Opcional: Verificar se o artifact_location está correto e corrigi-lo se necessário
    # (isso é mais avançado e geralmente é mais fácil recriar o experimento)
    # if experiment.artifact_location != tracking_uri:
    #     print("AVISO: O artifact_location do experimento existente é diferente do tracking_uri desejado.")

except AttributeError:
    print(f"Experimento '{EXPERIMENT_NAME}' não encontrado. Criando agora...")
    # O artifact_location também deve ser um URI.
    # Usar o mesmo URI do tracking garante que metadados e artefatos fiquem juntos.
    artifact_location = tracking_uri
    experiment_id = mlflow.create_experiment(EXPERIMENT_NAME, artifact_location=artifact_location)
    print(f"Experimento '{EXPERIMENT_NAME}' criado com ID: {experiment_id} e Artifact Location: {artifact_location}")

# 6. Defina o experimento que será usado para as próximas runs
mlflow.set_experiment(experiment_name=EXPERIMENT_NAME)


# 7. Agora, a run usará o tracking e o artifact URI corretos.
with mlflow.start_run(run_name="teste_com_uri_correto") as run:
    print(f"\n--- Iniciando Run ---")
    print(f"Run ID: {run.info.run_id}")
    print(f"Artifact URI para esta run: {run.info.artifact_uri}")

    features_list = [1, 2, 3, 4, 5, 6]
    arquivo_features_path = "features_para_logar.txt"
    with open(arquivo_features_path, "w") as f:
        f.write("\n".join(map(str, features_list)))
            
    mlflow.log_artifact(arquivo_features_path, artifact_path="model_config")
    print(f"Artefato '{arquivo_features_path}' logado em 'model_config'.")
    
    if os.path.exists(arquivo_features_path):
        os.remove(arquivo_features_path)

print("\nVerifique a pasta 'C:/palpiteiro/mlruns' e a UI do MLflow. O artefato deve estar lá.")
```

```python
import mlflow
import os

EXPERIMENT_NAME = "Modelo_B3_v4" # Nova versão para um teste limpo

# 1. Defina o caminho base ABSOLUTO.
CAMINHO_BASE_MLFLOW = r'C:\palpiteiro\mlruns_v4'
os.makedirs(CAMINHO_BASE_MLFLOW, exist_ok=True)

# 2. Processe o caminho ANTES de passá-lo para o f-string.
caminho_processado = os.path.abspath(CAMINHO_BASE_MLFLOW).replace('\\', '/')

# 3. Construa o Tracking URI.
tracking_uri = f"file:///{caminho_processado}"

# 4. Defina o Tracking URI
mlflow.set_tracking_uri(tracking_uri)
print(f"MLflow Tracking URI definido para: {mlflow.get_tracking_uri()}")

# 5. Verifique se o experimento existe. Se não, crie-o.
#    A principal mudança é NÃO especificar o artifact_location na criação.
#    Deixe o MLflow usar seu padrão, que geralmente é uma subpasta do tracking_uri.
try:
    experiment = mlflow.get_experiment_by_name(EXPERIMENT_NAME)
    if experiment is None:
        raise AttributeError
    
    experiment_id = experiment.experiment_id
    print(f"Experimento '{EXPERIMENT_NAME}' já existe (ID: {experiment_id}).")
    print(f"  Artifact Location existente: {experiment.artifact_location}")

except AttributeError:
    print(f"Experimento '{EXPERIMENT_NAME}' não encontrado. Criando agora...")
    # --- MUDANÇA CRUCIAL ---
    # NÃO passe artifact_location. Deixe o MLflow criar o padrão.
    # O padrão será um caminho relativo dentro do tracking_uri, que é o comportamento desejado.
    experiment_id = mlflow.create_experiment(EXPERIMENT_NAME)
    
    # Vamos verificar o artifact_location que o MLflow criou por padrão
    novo_experimento = mlflow.get_experiment(experiment_id)
    print(f"Experimento '{EXPERIMENT_NAME}' criado com ID: {experiment_id}")
    print(f"  Artifact Location padrão criado pelo MLflow: {novo_experimento.artifact_location}")


# 6. Defina o experimento que será usado para as próximas runs
mlflow.set_experiment(experiment_name=EXPERIMENT_NAME)

# 7. Execute a run.
with mlflow.start_run(run_name="teste_com_uri_correto_v4") as run:
    print(f"\n--- Iniciando Run ---")
    print(f"Run ID: {run.info.run_id}")
    # Este URI agora deve apontar para uma subpasta com o ID do experimento
    print(f"Artifact URI para esta run: {run.info.artifact_uri}")

    features_list = [10, 20, 30, 40, 50]
    arquivo_features_path = "features_para_logar.txt"
    with open(arquivo_features_path, "w") as f:
        f.write("\n".join(map(str, features_list)))
            
    mlflow.log_artifact(arquivo_features_path, artifact_path="model_config")
    print(f"Artefato '{arquivo_features_path}' logado em 'model_config'.")
    
    if os.path.exists(arquivo_features_path):
        os.remove(arquivo_features_path)

print(f"\nVerifique a pasta '{CAMINHO_BASE_MLFLOW}'. Você deve ver uma subpasta com o ID do experimento (ex: '1', '2', etc.), e dentro dela, as pastas das runs.")
```
